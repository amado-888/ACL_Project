### Step 1: Audit missing values before cleaning

`qualifying_df.info()` above reports `q1` as 10494/10494 non-null. That is misleading: `\N` is stored as a normal string, so pandas does not see it as missing.
We count two kinds of missing value separately:
- **sentinel**: the literal string `\N`
- **empty**: a truly empty cell, which pandas already reads as `NaN`

```python
def missing_audit(df):
    audit = pd.DataFrame({
        "sentinel": df.astype(str).eq(SENTINEL).sum(),
        "empty": df.isna().sum(),
    })
    audit["total_missing"] = audit["sentinel"] + audit["empty"]
    audit["pct_missing"] = (audit["total_missing"] / len(df) * 100).round(1)
    return audit[audit["total_missing"] > 0]

q_audit = missing_audit(qualifying)
q_audit
```

```python
fig, ax = plt.subplots(figsize=(8, 3))
ax.barh(q_audit.index, q_audit["sentinel"], color=BLUE, label="sentinel '\\N'")
ax.barh(q_audit.index, q_audit["empty"], left=q_audit["sentinel"], color=ORANGE, label="empty cell")
for i, total in enumerate(q_audit["total_missing"]):
    ax.text(total + 80, i, f"{total:,} ({q_audit['pct_missing'].iloc[i]}%)", va="center")
ax.set_xlim(0, len(qualifying))
ax.invert_yaxis()
ax.set_xlabel(f"Rows (table has {len(qualifying):,})")
ax.set_title("Qualifying: missing lap times before cleaning")
ax.legend(loc="upper right")
sns.despine()
plt.show()
```

**Observations**
- Only `q1`, `q2` and `q3` have missing values, and the amount grows from Q1 to Q3 (1.5%, 44%, 65%).
- There are **two different codes** for missing: most rows use `\N`, but a few rows have a truly empty cell. Below we check where the empty cells come from.

```python
empty_rows = qualifying[qualifying[["q2", "q3"]].isna().any(axis=1)]
empty_rows.merge(races[["raceId", "year", "name"]], on="raceId").groupby(["raceId", "year", "name"]).size()
```

All the empty cells come from 5 recent races (2023 Abu Dhabi and four races in 2024). This looks like a change in how the newest data was exported, not a different meaning, so `\N` and empty cells will both be turned into one `NaN`.

### Step 2: Why are q2/q3 missing? Random or by design?

Hypothesis: the missing values come from the **qualifying format** used in each era, not from data errors.
If that is true, the share of missing values should change sharply at the years when the format changed.

```python
def is_missing(s):
    return s.isna() | s.astype(str).eq(SENTINEL)

q_year = qualifying.merge(races[["raceId", "year", "round"]], on="raceId")
missing_by_year = (
    q_year.assign(**{c: is_missing(q_year[c]) for c in ["q1", "q2", "q3"]})
          .groupby("year")[["q1", "q2", "q3"]].mean() * 100
)

fig, ax = plt.subplots(figsize=(10, 4))
for col, color in zip(["q1", "q2", "q3"], [BLUE, ORANGE, AQUA]):
    ax.plot(missing_by_year.index, missing_by_year[col], color=color, linewidth=2, marker="o", markersize=4, label=col)
ax.axvline(2005.5, color="grey", linestyle="--", linewidth=1)
ax.text(1999.5, 50, "single-session\nqualifying\n(only one lap time)", ha="center", color="grey")
ax.text(2015, 85, "knockout Q1 / Q2 / Q3\n(2006 onwards)", ha="center", color="grey")
ax.set_ylim(0, 105)
ax.set_xlabel("Season")
ax.set_ylabel("% of rows missing a time")
ax.set_title("Share of qualifying rows with no lap time, per season")
ax.legend(loc="center left", bbox_to_anchor=(1, 0.5))
sns.despine()
plt.show()

missing_by_year.round(1).T
```

**Interpretation (the hypothesis holds):**
- **1994–2004:** `q2`/`q3` are 100% missing. Qualifying was one session, so only one time exists. This is *structural* missingness, not an error.
- **2005:** two-session aggregate qualifying was used for the first 6 rounds only, so `q2` exists for part of the season.
- **2006 onwards (knockout):** about 25–30% miss `q2` and about 50% miss `q3`. That matches the slowest drivers being eliminated after Q1 and Q2.
- `q1` is missing in only ~1.5% of rows in every era. Those are drivers who set no time at all (crash, mechanical failure, excluded).

If knockout elimination is the cause, then in the 2006+ era having a Q2/Q3 time should depend on the final qualifying **position**, with a sharp drop at the cut-off.

```python
knockout = q_year[q_year["year"] >= 2006]
has_time_by_pos = (
    knockout.assign(has_q2=~is_missing(knockout["q2"]), has_q3=~is_missing(knockout["q3"]))
            .groupby("position")[["has_q2", "has_q3"]].mean() * 100
)

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(has_time_by_pos.index, has_time_by_pos["has_q2"], color=ORANGE, linewidth=2, marker="o", markersize=5, label="has a q2 time")
ax.plot(has_time_by_pos.index, has_time_by_pos["has_q3"], color=AQUA, linewidth=2, marker="o", markersize=5, label="has a q3 time")
ax.set_xticks(has_time_by_pos.index)
ax.set_xlabel("Final qualifying position")
ax.set_ylabel("% of rows")
ax.set_title("Knockout era (2006+): who has a Q2 / Q3 time, by qualifying position")
ax.legend()
sns.despine()
plt.show()

print("Qualifying positions of rows with a missing q1:")
print(qualifying.loc[is_missing(qualifying["q1"]), "position"].describe()[["count", "min", "50%", "max"]])
```

**Interpretation:**
- P1–P8 almost always have a Q3 time and P1–P14 almost always have a Q2 time. After the cut-off (P11 for Q3, P15–P17 for Q2) the share falls to 0. So a missing `q2`/`q3` means **"eliminated earlier"**, not "unknown".
- The Q2 cut-off is spread over P15–P17 because the grid size (20–24 cars) and the number of drivers eliminated changed between seasons.
- The dip at P9–P10 (77–93%) is drivers who **reached Q3 but set no lap**, for example to save tyres for the race (common in 2011–2015) or after a crash. A driver without a time is ranked last in the Q3 group, which is why they collect at P9/P10. Here a missing `q3` is not "eliminated", so it is a known limitation of reading `q3` alone.
- Rows with a missing `q1` are mostly at the back (median P20), which is consistent with "set no time".

**Decision and why:**
- Replace every `\N` / empty cell with `NaN`. Do **not** fill with `0` (it would look like the fastest lap ever) and do **not** copy `q1` into `q2`/`q3` (that would hide that the driver was eliminated).
- Do **not drop** rows with missing `q2`/`q3`. That would delete every pre-2006 race and every driver eliminated early, and leave mostly front-runners, which biases the data toward the podium class we are trying to predict.
- Any imputation for modelling is left until after the train/validation/test split, so values from test years cannot leak into training.

```python
lap_cols = ["q1", "q2", "q3"]
qualifying[lap_cols] = qualifying[lap_cols].replace(SENTINEL, np.nan)

assert not qualifying.astype(str).eq(SENTINEL).any().any(), "a \\N sentinel is still present"
qualifying[lap_cols].isna().sum()
```

### Step 3: Semantic typing (lap-time strings to numeric seconds)

All non-missing values are first checked against the pattern `m:ss.sss`, so the parser cannot silently misread a different format.
We use **seconds** (not milliseconds) to stay consistent with `pit_stops["duration_seconds"]` in the Scheduling & Timing section.

The parser also accepts strings without minutes (e.g. `'1.430'`), because the sprint table needs that too.

```python
LAP_PATTERN = r"^\d+:\d{2}\.\d{3}$"
for col in lap_cols:
    values = qualifying[col].dropna()
    print(f"{col}: {(~values.str.match(LAP_PATTERN)).sum()} values not in m:ss.sss format")

def to_seconds(t):
    """'1:26.572' -> 86.572, '1.430' -> 1.43, NaN -> NaN"""
    if pd.isna(t):
        return np.nan
    minutes, _, seconds = t.rpartition(":")
    return (float(minutes) if minutes else 0.0) * 60 + float(seconds)
```

```python
for col in lap_cols:
    qualifying[f"{col}_sec"] = qualifying[col].map(to_seconds)
    # Parsing must not create or remove any missing values
    assert (qualifying[f"{col}_sec"].isna() == qualifying[col].isna()).all()

qualifying[["q1", "q1_sec", "q2", "q2_sec", "q3", "q3_sec"]].head()
```

```python
# The string versions are no longer needed
qualifying = qualifying.drop(columns=lap_cols)
qualifying.info()
```

### Step 4: Check the parsed times for impossible values

Raw seconds cannot be compared across races, because circuits have very different lengths. A 90 s lap is fast at Spa but very slow at the Red Bull Ring.
So each driver's best time is measured against the **pole time of the same race**:

`gap_to_pole_pct = (driver best time / pole time - 1) * 100`

The best time is the minimum of q1/q2/q3, not the last session the driver took part in. In 2006–2009, Q3 was run with race fuel, so Q3 laps were often **slower** than the same driver's Q2 lap.

```python
def add_pace_features(df):
    df["best_quali_sec"] = df[["q1_sec", "q2_sec", "q3_sec"]].min(axis=1)
    pole_time = df.groupby("raceId")["best_quali_sec"].transform("min")
    df["gap_to_pole_pct"] = (df["best_quali_sec"] / pole_time - 1) * 100
    return df

qualifying = add_pace_features(qualifying)

fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(qualifying["gap_to_pole_pct"].dropna(), bins=200, color=BLUE)
ax.set_yscale("log")
ax.set_xlabel("Gap to pole (%)")
ax.set_ylabel("Rows (log scale)")
ax.set_title("Gap to pole before outlier handling")
sns.despine()
plt.show()

(qualifying.merge(races[["raceId", "year", "name"]], on="raceId")
           .nlargest(6, "gap_to_pole_pct")
           [["raceId", "year", "name", "driverId", "position", "q1_sec", "best_quali_sec", "gap_to_pole_pct"]])
```

**Outlier rule and why:**
- One row (1995, `16:42.640`, about 1002 s) is **~9x the pole time**. No real flying lap can be that slow. It is either a recording error or a car crawling back to the pits.
- The next slowest laps are 30–45% off pole. Those are real: wet or red-flagged sessions where some drivers never got a clean lap.
- So we **do not** use the official 107% rule here. That rule was not enforced in every season, and it would delete real wet-session laps.
- Rule used: a best lap more than **100% slower than pole** (over twice the pole time) is not a valid lap, so it is set to `NaN`. This removes exactly one value.

```python
MAX_GAP_PCT = 100
outlier = qualifying["gap_to_pole_pct"] > MAX_GAP_PCT
print("Rows flagged as impossible lap times:", outlier.sum())

qualifying.loc[outlier, ["q1_sec", "q2_sec", "q3_sec"]] = np.nan
qualifying = add_pace_features(qualifying)

qualifying[["q1_sec", "q2_sec", "q3_sec", "best_quali_sec", "gap_to_pole_pct"]].describe().round(2)
```

### Step 5: Effect of normalising the pace (raw seconds vs gap to pole)

Below we compare both versions against the final qualifying position. A useful pace measure should rise steadily with position.

```python
top20 = qualifying[qualifying["position"] <= 20]

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
sns.boxplot(data=top20, x="position", y="best_quali_sec", color=BLUE, showfliers=False, linewidth=1, ax=axes[0])
axes[0].set_title("Before: raw best lap (seconds)")
axes[0].set_ylabel("Best qualifying lap (s)")
sns.boxplot(data=top20, x="position", y="gap_to_pole_pct", color=AQUA, showfliers=False, linewidth=1, ax=axes[1])
axes[1].set_title("After: gap to pole (%)")
axes[1].set_ylabel("Gap to pole (%)")
for ax in axes:
    ax.set_xlabel("Final qualifying position")
sns.despine()
plt.tight_layout()
plt.show()

print("Spearman correlation with qualifying position:")
print(qualifying[["position", "best_quali_sec", "gap_to_pole_pct"]].corr(method="spearman")["position"].drop("position").round(3))
```

**Interpretation:**

Before cleaning, we check what each sentinel means by comparing it with the classification code (`positionText`) and the number of laps completed.

```python
sprint_check = sprint_results.assign(
    position_missing=sprint_results["position"].eq(SENTINEL),
    time_missing=sprint_results["milliseconds"].eq(SENTINEL),
    fastest_lap_missing=sprint_results["fastestLap"].eq(SENTINEL),
)
print("positionText of rows with position = \\N:", sprint_check.loc[sprint_check["position_missing"], "positionText"].value_counts().to_dict())
print("\nposition missing vs finish time missing:")
print(pd.crosstab(sprint_check["position_missing"], sprint_check["time_missing"]))
print("\nLaps completed by rows with no fastest lap:", sprint_check.loc[sprint_check["fastest_lap_missing"], "laps"].value_counts().to_dict())
print("\nThe 'time' column holds two kinds of value:")
print(sprint_results.loc[sprint_results["positionOrder"] <= 3, ["raceId", "positionOrder", "time", "milliseconds"]].head(3))
```

**What each sentinel means:**
- `position = \N` (15 rows): the driver was **not classified** (`positionText` R = retired, N = not classified, W = withdrawn). `positionOrder` still ranks them, so it is kept as is.
- `time` / `milliseconds = \N` (20 rows): **no finish time**. That covers the 15 unclassified drivers plus 5 who retired late but were still classified (they had covered enough distance).
- `fastestLap` / `fastestLapTime = \N` (9 rows): 6 drivers completed **0 laps** and 3 retired after 1–2 laps, so no valid timed lap was recorded.
- `grid = 0` (11 rows): **pit-lane start**. Left as `0`, a model would read it as "better than pole", so it becomes `NaN` with a separate flag.
- The `time` string has **two meanings in one column**: the winner's total time (`25:38.426`) and everyone else's gap (`+1.430`). It is replaced with two clear numeric columns built from `milliseconds`, which holds the total time for every finisher.

All these values are missing for a reason, so they become `NaN` (no fill, no drop).

```python
sprint_results = sprint_results.replace(SENTINEL, np.nan)

# Hidden sentinel: grid 0 = pit-lane start
sprint_results["sprint_pitlane_start"] = sprint_results["grid"].eq(0)
sprint_results["grid"] = sprint_results["grid"].replace(0, np.nan).astype("Int64")

# Integer columns that were stored as text because of '\N'
for col in ["position", "fastestLap"]:
    sprint_results[col] = pd.to_numeric(sprint_results[col]).astype("Int64")
sprint_results["sprint_classified"] = sprint_results["position"].notna()

# Durations to seconds
sprint_results["race_time_sec"] = pd.to_numeric(sprint_results["milliseconds"]) / 1000
sprint_results["gap_to_winner_sec"] = (
    sprint_results["race_time_sec"] - sprint_results.groupby("raceId")["race_time_sec"].transform("min")
)
sprint_results["fastestLapTime_sec"] = sprint_results["fastestLapTime"].map(to_seconds)

assert not sprint_results.astype(str).eq(SENTINEL).any().any(), "a \\N sentinel is still present"
```

```python
# Check: the gap computed from milliseconds must match the original '+x.xxx' strings
gap_rows = sprint_results["time"].str.startswith("+", na=False)
gap_from_string = sprint_results.loc[gap_rows, "time"].str.lstrip("+").map(to_seconds)
assert np.allclose(gap_from_string, sprint_results.loc[gap_rows, "gap_to_winner_sec"], atol=0.002)
print(f"{gap_rows.sum()} gap strings match the gap computed from milliseconds")

sprint_results = sprint_results.drop(columns=["time", "milliseconds", "fastestLapTime"])
sprint_results.info()
```
