```python
drivers_clean = drivers_df.copy()
driver_standings_clean = driver_standings_df.copy()
constructors_clean = constructors_df.copy()
constructor_standings_clean = constructor_standings_df.copy()
circuits_clean = circuits_df.copy()
```

```python
drivers_clean = drivers_clean.replace(r"\N", np.nan)
```

```python
drivers_clean.isna().sum()
```

```python
plt.figure(figsize=(8, 4))
sns.barplot(x=drivers_clean.isna().sum().index, y=drivers_clean.isna().sum().values)
plt.title("Missing Values in Drivers Table (Before Dropping)")
plt.ylabel("Count of Missing (NaN)")
plt.xticks(rotation=45)
plt.show()
```


**As we can see since we have a total of 861 entries per column in the drivers table we have a lot of nulls in both number and code Where code represents the first three letters of the driver's surname, which does not affect predicting the podium, so it can be safely droppedand the number column, which is the driver's permanent racing car number, has no effect on predicting the podium, so it can be safely dropped. Inaddition to the url that can be dropped also it has nos significance on predicting the podium**


```python
drivers_clean=drivers_clean.drop(columns = ["number", "code", "url"])
```

**Now after dropping unnecessary columns we will convert the dob to a valid datetime object in the drivers table its in the following format (DD/MM/YYY)**

```python
drivers_clean['dob'] = pd.to_datetime(drivers_clean['dob'], format='%Y-%m-%d')
```

```python
# 1. Primary Key Uniqueness & Non-null checks
assert drivers_clean['driverId'].is_unique and drivers_clean['driverId'].notna().all(), "driverId PK violation!"
assert constructors_clean['constructorId'].is_unique and constructors_clean['constructorId'].notna().all(), "constructorId PK violation!"
assert circuits_clean['circuitId'].is_unique and circuits_clean['circuitId'].notna().all(), "circuitId PK violation!"
assert driver_standings_clean['driverStandingsId'].is_unique and driver_standings_clean['driverStandingsId'].notna().all(), "driverStandingsId PK violation!"
assert constructor_standings_clean['constructorStandingsId'].is_unique and constructor_standings_clean['constructorStandingsId'].notna().all(), "constructorStandingsId PK violation!"

print("All primary keys have been successfully verified! No violations found.")
```

```python
# 2. Foreign Key Reconciliation
assert driver_standings_clean['driverId'].isin(drivers_clean['driverId']).all(), "FK mismatch: driver_standings -> drivers"
assert constructor_standings_clean['constructorId'].isin(constructors_clean['constructorId']).all(), "FK mismatch: constructor_standings -> constructors"

print("All Foreign keys have been successfully verified! No violations found.")
```

```python
# Extracting unique nationalities and countries from the circuits and drivers columns
unique_nationalities = drivers_clean['nationality'].unique()
unique_circuits = circuits_clean['country'].unique()

print(f"Nationalities: {unique_nationalities}")
print(f"circuits: {unique_circuits}")
```

**After examining the unique countries and nationalities we encountered the following** 
1. We have two countries mapping to the same nationality (USA and United States), so we will standardize to USA.
2. Not all countries hosted a Grand Prix for all nationalities of the drivers, so when adding the driver country feature some values will be NaN (meaning these countries did not host a Grand Prix) 

```python
#Standardizing to USA
circuits_clean['country'] = circuits_clean['country'].replace('United States', 'USA')
```

```python
# Mapping driver nationality (demonym) to circuit country
nat_to_country = {
    'British': 'UK',
    'German': 'Germany',
    'East German': 'Germany', 
    'Spanish': 'Spain',
    'Japanese': 'Japan',
    'French': 'France',
    'Brazilian': 'Brazil',
    'Italian': 'Italy',
    'Australian': 'Australia',
    'Austrian': 'Austria',
    'American': 'USA',       
    'Dutch': 'Netherlands',
    'Portuguese': 'Portugal',
    'Canadian': 'Canada',
    'Indian': 'India',
    'Hungarian': 'Hungary',
    'Argentine': 'Argentina',
    'Argentinian ': 'Argentina',
    'Malaysian': 'Malaysia',
    'Swiss': 'Switzerland',
    'Belgian': 'Belgium',
    'Monegasque': 'Monaco',
    'Swedish': 'Sweden',
    'Mexican': 'Mexico',
    'South African': 'South Africa',
    'Russian': 'Russia',
    'Chinese': 'China',
    'American-Italian': 'USA', 
    'Argentine-Italian': 'Argentina'
}
```

```python
drivers_clean['driver_country'] = drivers_clean['nationality'].map(nat_to_country)
drivers_clean.head()
```

```python
# Dropping unnecessary data that won't affect the podium prediction
constructors_clean = constructors_clean.drop(columns=['url'])
circuits_clean = circuits_clean.drop(columns=['url'])
```
